```python
from IPython.display import display, HTML

logo_html = """
<div style="font-family: sans-serif; background: #222; color: #fff; text-align: center; padding: 20px; border-radius: 8px; max-width: 850px; margin: 0 auto; user-select: none;">
    <h2 style="margin-top: 0;">🐢 Mini LOGO Turtle (Traços Calibrados)</h2>

    <div style="display: flex; justify-content: center; align-items: flex-start; gap: 20px; margin: 20px 0;">

        <!-- Painel Lateral de Controle -->
        <div style="display: flex; flex-direction: column; gap: 10px; background: #333; padding: 15px; border-radius: 6px; width: 160px; box-shadow: inset 0 0 10px rgba(0,0,0,0.5);">
            <span style="font-weight: bold; font-size: 14px; color: #aaa; text-transform: uppercase;">Movimento</span>
            <button onclick="controlar('pf')" style="padding: 10px; background: #00ffcc; border: none; border-radius: 4px; font-weight: bold; cursor: pointer; color: #111;">Frente (+50)</button>

            <span style="font-weight: bold; font-size: 14px; color: #aaa; text-transform: uppercase; margin-top: 10px;">Rotação</span>
            <button onclick="controlar('pe')" style="padding: 10px; background: #ffaa00; border: none; border-radius: 4px; font-weight: bold; cursor: pointer; color: #111;">Esq. (↰ 45°)</button>
            <button onclick="controlar('pd')" style="padding: 10px; background: #ffaa00; border: none; border-radius: 4px; font-weight: bold; cursor: pointer; color: #111;">Dir. (↱ 45°)</button>

            <!-- Botão de Ativar/Desativar Caneta -->
            <span style="font-weight: bold; font-size: 14px; color: #aaa; text-transform: uppercase; margin-top: 10px;">Caneta</span>
            <button id="btnCaneta" onclick="alternarCaneta()" style="padding: 10px; background: #4caf50; border: none; border-radius: 4px; font-weight: bold; cursor: pointer; color: #fff;">✏️ Desenhando</button>

            <!-- Seleção de Cores da Caneta -->
            <span style="font-weight: bold; font-size: 14px; color: #aaa; text-transform: uppercase; margin-top: 10px;">Cor da Linha</span>
            <div style="display: grid; grid-template-columns: repeat(3, 1fr); gap: 5px;">
                <button onclick="mudarCor('rgb(0,255,204)')" style="background: rgb(0,255,204); width: 35px; height: 35px; border: 2px solid white; border-radius: 4px; cursor: pointer;"></button>
                <button onclick="mudarCor('rgb(255,0,85)')" style="background: rgb(255,0,85); width: 35px; height: 35px; border: none; border-radius: 4px; cursor: pointer;"></button>
                <button onclick="mudarCor('rgb(255,234,0)')" style="background: rgb(255,234,0); width: 35px; height: 35px; border: none; border-radius: 4px; cursor: pointer;"></button>
                <button onclick="mudarCor('rgb(30,144,255)')" style="background: rgb(30,144,255); width: 35px; height: 35px; border: none; border-radius: 4px; cursor: pointer;"></button>
                <button onclick="mudarCor('rgb(255,255,255)')" style="background: rgb(255,255,255); width: 35px; height: 35px; border: none; border-radius: 4px; cursor: pointer;"></button>
                <button onclick="mudarCor('arcoiris')" style="background: linear-gradient(45deg, red, yellow, green, blue); width: 35px; height: 35px; border: none; border-radius: 4px; cursor: pointer;" title="Caneta Arco-Íris"></button>
            </div>

            <span style="font-weight: bold; font-size: 14px; color: #aaa; text-transform: uppercase; margin-top: 10px;">Sistema</span>
            <button onclick="controlar('limpe')" style="padding: 10px; background: #ff0055; border: none; border-radius: 4px; font-weight: bold; cursor: pointer; color: #fff;">Limpar Tela</button>
        </div>

        <!-- Área Única do Canvas -->
        <div style="position: relative; width: 600px; height: 400px;">
            <canvas id="canvasLogo" width="600" height="400" style="background: #111; border: 2px solid #444; border-radius: 4px; cursor: crosshair;"></canvas>
        </div>
    </div>

    <input type="text" id="console" placeholder="Comandos: pf 50, pd 90, sc (solta caneta), pc (pega caneta)"
           style="width: 550px; padding: 10px; font-size: 16px; background: #333; color: #0f0; border: none; font-family: monospace; border-radius: 4px;" autofocus>
</div>

<script>
var controlar;
var mudarCor;
var alternarCaneta;

(function() {
    const canvas = document.getElementById('canvasLogo');
    const ctx = canvas.getContext('2d');
    const consoleInput = document.getElementById('console');
    const btnCaneta = document.getElementById('btnCaneta');

    // Estado da tartaruga perfeitamente calibrado
    let t = {
        x: canvas.width / 2,
        y: canvas.height / 2,
        angulo: -Math.PI / 2, // Começa olhando para cima
        canetaBaixo: true,
        corAtual: 'rgb(0,255,204)',
        modoArcoiris: false,
        hueArcoiris: 0
    };

    let linhasSalvas = [];

    function desenharTartarugaGeometrica() {
        ctx.save();
        ctx.translate(t.x, t.y);
        ctx.rotate(t.angulo + Math.PI / 2); // Ajusta a rotação visual para bater com o ângulo matemático

        // Patas e rabo mudam de cor se a caneta estiver levantada
        ctx.fillStyle = t.canetaBaixo ? 'rgb(34,139,34)' : 'rgb(100,100,100)';
        ctx.beginPath(); ctx.arc(-18, -15, 6, 0, 2*Math.PI); ctx.fill();
        ctx.beginPath(); ctx.arc(18, -15, 6, 0, 2*Math.PI); ctx.fill();
        ctx.beginPath(); ctx.arc(-16, 18, 5, 0, 2*Math.PI); ctx.fill();
        ctx.beginPath(); ctx.arc(16, 18, 5, 0, 2*Math.PI); ctx.fill();

        ctx.beginPath(); ctx.moveTo(-4, 22); ctx.lineTo(0, 32); ctx.lineTo(4, 22); ctx.closePath(); ctx.fill();

        // Cabeça
        ctx.fillStyle = t.canetaBaixo ? 'rgb(50,205,50)' : 'rgb(140,140,140)';
        ctx.beginPath(); ctx.arc(0, -26, 10, 0, 2*Math.PI); ctx.fill();

        ctx.fillStyle = '#000';
        ctx.beginPath(); ctx.arc(-4, -29, 2, 0, 2*Math.PI); ctx.fill();
        ctx.beginPath(); ctx.arc(4, -29, 2, 0, 2*Math.PI); ctx.fill();

        // Casco Principal
        ctx.fillStyle = t.canetaBaixo ? 'rgb(46,139,87)' : 'rgb(80,80,80)';
        ctx.strokeStyle = t.canetaBaixo ? 'rgb(15,76,36)' : 'rgb(50,50,50)';
        ctx.lineWidth = 2;
        ctx.beginPath(); ctx.arc(0, 0, 20, 0, 2*Math.PI); ctx.fill(); ctx.stroke();

        ctx.beginPath(); ctx.arc(0, 0, 10, 0, 2*Math.PI); ctx.stroke();
        ctx.restore();
    }

    // Renderizador mestre corrigido
    function renderizarTela() {
        ctx.clearRect(0, 0, canvas.width, canvas.height);

        ctx.lineWidth = 3;
        ctx.lineCap = "round"; // Deixa as junções das linhas mais suaves

        // Renderiza cada linha salva individualmente sem erros de coordenadas
        linhasSalvas.forEach(linha => {
            ctx.strokeStyle = linha.cor;
            ctx.beginPath();
            ctx.moveTo(linha.x1, linha.y1);
            ctx.lineTo(linha.x2, linha.y2);
            ctx.stroke();
        });

        desenharTartarugaGeometrica();
    }

    function inicializar() {
        linhasSalvas = [];
        t.x = canvas.width / 2;
        t.y = canvas.height / 2;
        t.angulo = -Math.PI / 2;
        t.canetaBaixo = true;
        atualizarInterfaceCaneta();
        renderizarTela();
    }

    function atualizarInterfaceCaneta() {
        if (t.canetaBaixo) {
            btnCaneta.innerText = "✏️ Desenhando";
            btnCaneta.style.background = "#4caf50";
        } else {
            btnCaneta.innerText = "❌ Sem Rastro";
            btnCaneta.style.background = "#777";
        }
    }

    alternarCaneta = function() {
        t.canetaBaixo = !t.canetaBaixo;
        atualizarInterfaceCaneta();
        renderizarTela();
    };

    mudarCor = function(novaCor) {
        const botoes = document.querySelectorAll('div > button');
        botoes.forEach(b => b.style.border = 'none');
        event.target.style.border = '2px solid white';

        if (novaCor === 'arcoiris') {
            t.modoArcoiris = true;
        } else {
            t.modoArcoiris = false;
            t.corAtual = novaCor;
        }
    };

    function moverFrente(passos) {
        let novoX = t.x + passos * Math.cos(t.angulo);
        let novoY = t.y + passos * Math.sin(t.angulo);

        if (t.canetaBaixo) {
            let corDaLinha = t.corAtual;
            if (t.modoArcoiris) {
                corDaLinha = `hsl(${t.hueArcoiris}, 100%, 50%)`;
                t.hueArcoiris = (t.hueArcoiris + 12) % 360;
            }
            // Guarda as coordenadas limpas
            linhasSalvas.push({
                x1: t.x, y1: t.y,
                x2: novoX, y2: novoY,
                cor: corDaLinha
            });
        }

        t.x = novoX;
        t.y = novoY;
        renderizarTela();
    }

    controlar = function(acao) {
        if (acao === 'pf') moverFrente(50);
        if (acao === 'pd') { t.angulo += (45 * Math.PI) / 180; renderizarTela(); }
        if (acao === 'pe') { t.angulo -= (45 * Math.PI) / 180; renderizarTela(); }
        if (acao === 'limpe') inicializar();
    };

    // Mover a tartaruga por clique sem rabiscar
    canvas.addEventListener('click', function(e) {
        const rect = canvas.getBoundingClientRect();
        t.x = e.clientX - rect.left;
        t.y = e.clientY - rect.top;
        renderizarTela();
    });

    function interpretar(comandoTexto) {
        const partes = comandoTexto.toLowerCase().trim().split(/\s+/);
        let i = 0;

        while (i < partes.length) {
            let cmd = partes[i];
            let val = parseFloat(partes[i+1]);

            if (cmd === 'pf' || cmd === 'para_frente') {
                moverFrente(val);
                i += 2;
            }
            else if (cmd === 'pd' || cmd === 'para_direita') {
                t.angulo += (val * Math.PI) / 180;
                i += 2;
            }
else if (cmd === 'pe' || cmd === 'para_esquerda') {
t.angulo -= (val * Math.PI) / 180;
i += 2;
}
else if (cmd === 'sc' || cmd === 'solta_caneta' || cmd === 'un') {
t.canetaBaixo = false;
atualizarInterfaceCaneta();
i += 1;
}
else if (cmd === 'pc' || cmd === 'pega_caneta' || cmd === 'usecaneta') {
t.canetaBaixo = true;
atualizarInterfaceCaneta();
i += 1;
}
else if (cmd === 'limpe') {
inicializar();
return;
}
else {
i++;
}
}
renderizarTela();
}
consoleInput.addEventListener('keydown', (e) => {
if (e.key === 'Enter') {
interpretar(consoleInput.value);
consoleInput.value = '';
}
});
inicializar();
})();

"""
display(HTML(logo_html))
```
